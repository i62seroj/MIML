```python
import sys
from pathlib import Path

sys.path.insert(
    0,
    str(Path.home() / "GitHub" / "MIML" / "src")
)
output_path = str(Path.home() / "GitHub" / "MIML" / "src" / "miml" / "tutorial")

from miml.data import load_dataset
#dataset = load_dataset("miml_birds.arff", from_library=True)

# Split dataset into train and test partitions
#dataset_train, dataset_test = dataset.split_dataset(train_percentage=0.8, seed=7)
# Birds
birds = load_dataset("miml_birds.arff", from_library=True)

birds_train, birds_test = birds.split_dataset(
    train_percentage=0.8,
    seed=7
)

# Text
text = load_dataset("miml_text_data.arff", from_library=True)

text_train, text_test = text.split_dataset(
    train_percentage=0.8,
    seed=7
)

# Image
image = load_dataset("miml_image_data.arff", from_library=True)

image_train, image_test = image.split_dataset(
    train_percentage=0.8,
    seed=7
)

import miml

print(miml.__file__)
```

```python

import os
from miml.report import Report

from miml.classifier.miml.lazy.miml_br_knn import MIMLBRkNN
from miml.classifier.miml.lazy.miml_dgc import MIMLDGC
from miml.classifier.miml.lazy.miml_map_knn import MIMLMAPkNN
from miml.classifier.miml.lazy.miml_knn import MIMLkNN


datasets = {
    "birds": (birds_train, birds_test),
    #"text": (text_train, text_test),
    #"image": (image_train, image_test),
}


all_reports = {}

for dataset_name, (dataset_train, dataset_test) in datasets.items():

    print("\n" + "#" * 100)
    print(f"DATASET: {dataset_name}")
    print("#" * 100)

    classifiers = {
        "MIMLBRkNN": MIMLBRkNN(num_of_neighbours=10),
        "MIMLDGC": MIMLDGC(num_of_neighbours=10),
        "MIMLMAPkNN": MIMLMAPkNN(num_of_neighbours=10),
        "MIMLkNN": MIMLkNN(
            num_references=10,
            num_citers=10
        )
    }

    all_reports[dataset_name] = {}

    for clf_name, clf in classifiers.items():

        print("\n" + "=" * 80)
        print(clf_name)
        print("=" * 80)

        clf.build(dataset_train)

        predictions = clf.evaluate(dataset_test)
        probabilities = clf.predict_proba(dataset_test)

        report = Report(
            predictions,
            probabilities,
            dataset_test
        )

        report.calculate_metrics()

        all_reports[dataset_name][clf_name] = report

        report.to_string()
        report.to_csv(output_path + f"/{dataset_name}_{clf_name}.csv")

        print()
```
