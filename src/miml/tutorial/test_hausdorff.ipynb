# Import metrics, classifier and Hausdorff distances

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    hamming_loss,
)

# Add the project source directory to Python's path.
sys.path.insert(
    0,
    str(Path.home() / "GitHub" / "MIML" / "src"),
)

from miml.data import load_dataset
from miml.report import Report

from miml.classifier.miml.lazy.miml_dgc import MIMLDGC

from miml.core.average_hausdorff import AverageHausdorff
from miml.core.maximal_hausdorff import MaximalHausdorff
from miml.core.minimal_hausdorff import MinimalHausdorff
```

# Load datasets

```python
birds = load_dataset(
    "miml_birds.arff",
    from_library=True,
)

birds_train, birds_test = birds.split_dataset(
    train_percentage=0.8,
    seed=7,
)
```

# Run algorithm and save results in table

```python
classifier = MIMLDGC(metric=AverageHausdorff(), num_of_neighbours=10)
classifier.build(birds_train) 

average_distances = pd.DataFrame(classifier.D)

average_distances
```

```python

classifier = MIMLDGC(metric=MinimalHausdorff(), num_of_neighbours=10)
classifier.build(birds_train) 
minimal_distances = pd.DataFrame(classifier.D)

minimal_distances
```

```python

classifier = MIMLDGC(metric=MaximalHausdorff(), num_of_neighbours=10)
classifier.build(birds_train) 
maximal_distances = pd.DataFrame(classifier.D)

maximal_distances

```
