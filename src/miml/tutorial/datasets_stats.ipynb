```python
from miml.data import load_dataset


def dataset_statistics(dataset, name):
    """
    Obtain descriptive statistics for a MIML dataset.
    """

    n_bags = dataset.get_number_bags()

    return {
        "Dataset": name,
        "Características": dataset.get_number_features(),
        "Etiquetas": dataset.get_number_labels(),
        "Bolsas": n_bags,
        "Instancias/Bolsa": (
            dataset.get_number_instances() / n_bags
        ),
        "Densidad": dataset.density(),
        "Cardinalidad": dataset.cardinality(),
        "Distinct": dataset.distinct(),
    }


# ============================================================
# LOAD DATASETS
# ============================================================

birds = load_dataset(
    "miml_birds.arff",
    from_library=True,
)

text = load_dataset(
    "miml_text_data.arff",
    from_library=True,
)

image = load_dataset(
    "miml_image_data.arff",
    from_library=True,
)


# ============================================================
# CALCULATE STATISTICS
# ============================================================

statistics = [
    dataset_statistics(birds, "Birds"),
    dataset_statistics(text, "Text"),
    dataset_statistics(image, "Image"),
]


# ============================================================
# CREATE TABLE
# ============================================================

statistics_df = pd.DataFrame(statistics)

statistics_df
```
